```python
import pathlib 
import urllib.request
import datetime

class Adam: 
	species = 'homo sapiens'

	def __init__(self, verilmish_adi, adamin_yashi=None, adamin_cinsiyyeti=None, adam_yashayir=True):
		self.ad = verilmish_adi
		self.yash = adamin_yashi
		self.cins = adamin_cinsiyyeti
		self.yashayir = adam_yashayir
		self.adamin_folderi = None

	def adamin_profilini_yarat(self):
		profiller_folderi = pathlib.Path().cwd() / 'profiles'
		profiller_folderi.mkdir(parents=True, exist_ok=True)

		adamin_folderi = profiller_folderi / self.ad
		adamin_folderi.mkdir(parents=True, exist_ok=True)

		self.adamin_folderi = adamin_folderi


	def adamin_sheklini_yukle(self, sheklin_olcusu=400, ext = "jpg"):
		if self.adamin_folderi is None:
			self.adamin_profilini_yarat()
		
		url = f"https://api.dicebear.com/10.x/pixel-art/{ext}?seed={self.ad}&size={sheklin_olcusu}"

		now_str = datetime.datetime.today().strftime('%Y%m%d_%H%M%S')
		faylin_adi = self.adamin_folderi / f'{self.ad}_{now_str}.{ext}'

		urllib.request.urlretrieve(url, faylin_adi)
	
	def ol(self):
		self.yashayir = False

	def diril(self):
		self.yashayir = True

	def yashlan(self, nece_yash=1):
		if self.yash != None:
			self.yash = self.yash + nece_yash

adam1 = Adam(verilmish_adi='hesen', adamin_yashi=10, adamin_cinsiyyeti='K')
adam2 = Adam('ali')
```
