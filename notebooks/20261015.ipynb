```python
def adamin_sheklini_yukle(ext, adam='Felix', size=200):

	url = f"https://api.dicebear.com/10.x/pixel-art/{ext}?seed={adam}&size={size}"
	response = requests.get(url)

	img_path = IMAGES_DIR / f"{adam}-{size}.{ext}"
	with open(img_path, mode='wb') as f:
		f.write(response.content)
	return img_path

adamin_sheklini_yukle('svg', size=200)
```
