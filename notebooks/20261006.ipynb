```python
gunler = []

for i in range(0,7):
    day = datetime.date.today() + datetime.timedelta(days=i)
    gunler.append(day)

gunler
```

```python
for gun in gunler:
	heftenin_gunu = gun.weekday()

	# if heftenin_gunu == 5 or heftenin_gunu == 6:
	if heftenin_gunu in (5, 6):
		print('hefte sonudur')
	else:
		print('ish gunudur')
```
