```python
from sklearn.model_selection import train_test_split
from sklearn.datasets import fetch_california_housing

housing = fetch_california_housing()
X = housing['data']
y = housing['target']

X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.8)
X_valid, X_test, y_valid, y_test = train_test_split(X_test, y_test, train_size=0.5)
```

```python
X_train = torch.FloatTensor(X_train)
X_test = torch.FloatTensor(X_test)
X_valid = torch.FloatTensor(X_valid)
```

```python
means = X_train.mean(axis=0, keepdim=True)
std = X_train.std(axis=0, keepdim=True)
```

```python
X_train = (X_train - means) / std
X_test = (X_test - means) / std
X_valid = (X_valid - means) / std
```

```python
# y_train[:, np.newaxis]
y_train = torch.FloatTensor(y_train).view(-1,1)
y_test = torch.FloatTensor(y_test).view(-1,1)
y_valid = torch.FloatTensor(y_valid).view(-1,1)
```

```python
from torch import nn
```

```python
torch.manual_seed(123)

model = nn.Sequential(
	nn.Linear(8, 200),
	nn.ReLU(),
	nn.Linear(200, 100),
	nn.ReLU(),
	nn.Linear(100, 1),
)

sum([param.numel() for param in list(model.parameters())])
```

```python
n_epochs = 50
loss_function = nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.05)

for epoch in range(n_epochs):
	y_pred = model(X_train)
	loss = loss_function(y_pred, y_train)
	loss.backward()
	optimizer.step()
	optimizer.zero_grad()
	print(f"epoch: {epoch}/{n_epochs}, loss: {round(loss.item(),3)}")
```

```python
n_epochs = 50
loss_function = nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.05)

def train(model, loss_function, optimizer, X_train, y_train, n_epochs=50):
	for epoch in range(n_epochs):
		y_pred = model(X_train)
		loss = loss_function(y_pred, y_train)
		loss.backward()
		optimizer.step()
		optimizer.zero_grad()
		print(f"epoch: {epoch}/{n_epochs}, loss: {round(loss.item(),3)}")

train(model, loss_function, optimizer, X_train, y_train)
```
