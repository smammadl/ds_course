```python
# supervised learning
# unsupervised learning

from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt
import numpy as np
```

```python
blob_centers =np.array([
    [ 0.2,  2.3], 
	[-1.5,  2.3], 
	[-2.8,  1.8],
	[-2.8,  2.8], 
	[-2.8,  1.3]
])
blob_std = np.array(
    [0.4, 0.3, 0.1, 0.1, 0.1]
)

X, y = make_blobs(
	n_samples=500,
	n_features=2,
	centers=blob_centers,
	cluster_std=blob_std,
	random_state=0
)

plt.scatter(
	X[:, 0], 
	X[:, 1], 
	# c=y, 
	s=1, 
	cmap="Set1"
)
# plt.scatter(
# 	blob_centers[:, 0], 
# 	blob_centers[:, 1],
# 	marker='x',
# 	s=100
# )
plt.show()
```

```python
rng = np.random.default_rng(seed=122)
centroids = rng.choice(X, 5, replace=False, axis=0)

plt.scatter(
	X[:, 0], 
	X[:, 1], 
	# c=y, 
	s=1, 
	cmap="Set1"
)
plt.scatter(
	centroids[:, 0], 
	centroids[:, 1],
	marker='x',
	s=100
)
plt.show()
```

```python
from sklearn.metrics import pairwise_distances
from sklearn.metrics import pairwise_distances_argmin
# pairwise_distances(X, centroids).argmin(axis=1)
labels = pairwise_distances_argmin(X, centroids)

rng = np.random.default_rng(seed=122)
centroids = rng.choice(X, 5, replace=False, axis=0)

plt.scatter(
	X[:, 0], 
	X[:, 1], 
	c=labels, 
	s=1, 
	cmap="Set1"
)
plt.scatter(
	centroids[:, 0], 
	centroids[:, 1],
	marker='x',
	s=100
)
plt.show()
```

```python
centroids_new = []

for i in range(5):
	centroid_new = X[labels == i].mean(axis=0)
	centroids_new.append(centroid_new)

centroids = np.array(centroids_new)

plt.scatter(
	X[:, 0], 
	X[:, 1], 
	c=labels, 
	s=1, 
	cmap="Set1"
)
plt.scatter(
	centroids[:, 0], 
	centroids[:, 1],
	marker='x',
	s=100
)
plt.show()
```
