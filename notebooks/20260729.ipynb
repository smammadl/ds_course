```python
import numpy as np
import matplotlib.pyplot as plt


rng = np.random.default_rng(seed=42)
m = 2000  # number of instances

boylar = rng.normal(175, 10, size=(m,1))
yashlar = rng.uniform(18, 60, size=(m,1))
# maashlar = rng.exponential(scale=4, size=(m,1))
# maashlar = rng.chisquare(df=6, size=(m,1))
maashlar = rng.gamma(shape=3.5, scale=2, size=(m,1))

plt.hist(maashlar, bins=100)
plt.show()
```

```python
from sklearn.preprocessing import add_dummy_feature

X = np.hstack([boylar/100, yashlar/100])
X_b = add_dummy_feature(X)
# X_b = np.c_[X, np.ones((m, 1))]

y = maashlar/10

# X_b, y

y_cat = np.where(y<1.1, 0, 1)
y_cat
```

```python
X
```

```python
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import SGDClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression

log_reg = LogisticRegression()
log_reg.fit(X, y_cat.ravel())

dec_tree = DecisionTreeClassifier()
dec_tree.fit(X, y_cat.ravel())

sgd_cls = SGDClassifier()
sgd_cls.fit(X, y_cat.ravel())

knn_cls = KNeighborsClassifier()
knn_cls.fit(X, y_cat.ravel())

sv_cls = SVC()
sv_cls.fit(X, y_cat.ravel())
```

```python
log_reg.score(X,y_cat), dec_tree.score(X,y_cat), sgd_cls.score(X,y_cat), knn_cls.score(X,y_cat), sv_cls.score(X,y_cat)
```
