```python
import math
# import matplotlib.pyplot as plt

x = []
s = 0
for i in range(100):
	x.append(s)
	s += 0.2
y = [math.sin(i) for i in x]

# plt.figure(figsize = (2,2))
plt.plot(
	x,
	y, 
	marker = '*',
	linestyle = '--', 
	color = 'red'
	)
plt.show()
```

```python
import pandas as pd

df = pd.read_csv(r"C:\Users\user\Desktop\div\data\files\car-sales-extended-missing-data.csv")
x = df['Odometer (KM)'].to_list()
y = df['Price'].to_list()

# plt.plot(x,y, linestyle='', marker='.')
plt.scatter(x,y, s=3)
plt.xlabel('Odometer (KM)')
plt.ylabel('Price')
```

```python
df['Make'] = df['Make'].astype('category')
```

```python
df.plot.scatter(
	x = 'Odometer (KM)',
	y = 'Price',
	c = 'Make',
	marker = '.',
	cmap = 'tab10'
)
plt.show()
```

```python
df_g = df.groupby(by='Make', as_index=False)['Price'].mean().sort_values('Price', ascending=False)

df_g.plot.bar(
	x = "Make",
	y = 'Price',
	color=['red', 'green', 'blue', 'black'],
	legend=False
)
plt.xticks(rotation=0)
plt.ylabel('Price')
plt.show()
```

```python
df_g = df.groupby(by='Make')['Price'].mean()
df_g.plot.pie(autopct='%1.1f%%')
plt.show()
```
