```python
import math
pi = math.pi

def cevrenin_uzun_tap(r):
    global pi 
    pi = 3.14 
    return 2*pi*r 		# 2 * 3.14  * 5

def dairenin_sahesini_tap(r):
    return pi*(r**2)    # 3.14 * (5**2)

print(dairenin_sahesini_tap(r=5))
print(cevrenin_uzun_tap(r=5))
print(pi)
```

```python
import random

def random_list_yarat(min_sayi, max_sayi, element_sayi):

	bosh_list = []

	for i in range(element_sayi):
		random_sayi = random.randint(min_sayi, max_sayi)
		bosh_list.append(random_sayi)

	return bosh_list


l = random_list_yarat(5, 10, 20)
l
```
