```python
from PIL import Image # python image library
import datetime

def shekli_save_et(telebeler, sheklin_kohne_yolu):
	try:
		telebenin_adi = random.choice(telebeler)

		img = Image.open(sheklin_kohne_yolu)

		sheklin_yeni_yolu = fr"C:\Users\user\Desktop\notebooks\{telebenin_adi}.png"
		img.save(sheklin_yeni_yolu)
	except FileNotFoundError as e:
		erroru_yaz(e)
	except IndexError as e:
		erroru_yaz(e)
	except Exception as e:
		erroru_yaz(e)
	finally:
		return "shekli save etmek mumkun olmadi"

shekli_save_et([], r"C:\Users\user\Desktop\adam.png")
```
