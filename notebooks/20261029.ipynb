```python
import re

def pop_sutunlarini_tap(text):
	pattern = "\d{4} Population"
	# text = "World Population Percentage"

	match = re.search(pattern, text)
	return match is not None

pop_sutunlarini_tap("World Population Percentage")
```

```python
cols_to_show = ['Country'] + [col for col in list(df.columns) if pop_sutunlarini_tap(col)]
cols_to_show
```

```python
mask_pop = df['2022 Population'] > 1_000_000

mask_eur = df['Continent'] == 'Europe'

# df.loc[mask_pop, :]
df.loc[mask_pop & mask_eur, cols_to_show]
```

```python
magaza = {
	"kartof" : 0.8,
	"sogan" : 1.4,
	"alma" : 1.5,
	"banan" : 3
}

magaza.keys(), magaza.values()
```

```python
mehsullar = pd.Series(magaza.keys())
qiymetler = pd.Series(magaza.values())

df_magaza = pd.DataFrame(data={
	'mehsul' : mehsullar,
	'qiymet' : qiymetler,
})
mask_kartof = df_magaza['mehsul'] == 'kartof'
mask_sogan = df_magaza['mehsul'] == 'sogan'

odenis = df_magaza.loc[mask_kartof | mask_sogan, 'qiymet'] * 3
odenis.sum()
```
