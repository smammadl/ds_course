```python
class Adam:
	def __init__(self, a_adi=None, a_soyadi=None):
		self.__ad = a_adi
		self.__soyad = a_soyadi

	def adini_de(self):
		print(self.__ad)

	def adini_deyis(self, yeni_ad):
		self.__ad = yeni_ad

class Hekim(Adam):
	def __init__(self, a_adi=None, a_soyadi=None, h_sahesi=None):
		super().__init__(a_adi, a_soyadi)
		self.sahesi = h_sahesi

h1 = Hekim('lale', 'memmedova', 'terapevt')
h1.adini_deyis('jale')
h1.adini_de()
h1.sahesi
```

```python
class Qus:
	def __init__(self, uca_bilir = True):
		self.uca_bilir = uca_bilir

class Serce(Qus):
	def __init__(self, s_rengi, s_uca_bilir=True):
		super().__init__(uca_bilir=s_uca_bilir)
		self.reng = s_rengi

class Pinqvin(Qus):
	def __init__(self, p_rengi, p_novu):
		super().__init__(uca_bilir=False)
		self.reng = p_rengi
		self.__nov = p_novu

	@property
	def nov(self):
		return self.__nov

p1 = Pinqvin('boz', p_novu='E. sclateri')
p1.nov
```

```python
import math

class Cevre:
	cevrelerin_sayi = []

	def __init__(self, radius):
		self.r = radius
		Cevre.cevrelerin_sayi.append(self)

	@property
	def sahe(self):
		return math.pi * (self.r**2)
	
	@staticmethod
	def sahe_hesabla(r):
		return math.pi * (r**2)

	@classmethod
	def sayilarini_goster(cls):
		return len(cls.cevrelerin_sayi)

c1 = Cevre(10)
c2 = Cevre(20)
Cevre.sahe_hesabla(16)
Cevre.sayilarini_goster()
```

```python
class Oyuncu:
	def __init__(self, name, life):
		self.name = name
		self.life = life
		self.alive = True
		self.weapons = []
	
	def __gt__(self, other):
		return self.life > other.life

	def __ge__(self, other):
		return self.life >= other.life

	def __repr__(self):
		return f'<Oyuncu {self.name} {self.life}, {self.alive}, {self.weapons}>'

	def gullele(self, other):
		if other.life >= 10:
			other.life = other.life - 10
		else:
			other.die()

	def die(self):
		self.life = 0
		self.alive = False

	def kill(self, other):
		self.loot(other)
		other.die()

	def loot(self, other):
		self.weapons = self.weapons + other.weapons
		other.weapons = []

	def buy_weapon(self, weapon):
		self.weapons.append(weapon)

o1 = Oyuncu('xyz', 100)
o2 = Oyuncu('123', 90)
o1.buy_weapon('gun')
o2.buy_weapon('sword')
o1.kill(o2)
o1,o2
```
