```python
class Masin:
	def __init__(self, m_markasi, m_modeli, m_qiymeti, m_rengi=None, m_yurushu=None):
		self.marka = m_markasi
		self.model = m_modeli
		self.qiymet = m_qiymeti
		self.reng = m_rengi
		self.yurush = m_yurushu
		self.yaxshidir = True if self.qiymet < 10_000 else False
	
	def get(self, nece_km):
		self.yurush = self.yurush + nece_km

masin1 = Masin('Opel', 'Astra', 8700, m_rengi = 'goy', m_yurushu=320_550)
masin2 = Masin('Opel', 'Vectra', 40000)

print(masin1.yurush)
masin1.get(2)
print(masin1.yurush)
```

```python
class Adam:
	nov = 'homo sapiens'
	
	def __init__(self, a_adi, a_yashi=None, a_yashayir=True):
		self.ad = a_adi
		self.yash = a_yashi
		self.yashayir = a_yashayir
	def qocal(self, nece_yash):
		self.yash = self.yash + nece_yash
	def ol(self):
		self.yashayir = False

a1 = Adam(a_adi = 'ali', a_yashi=10)
a2 = Adam(a_adi = 'hesen', a_yashi=20)

isinstance(a1, Masin)
```

```python
class Sheher:
	def __init__(self, sh_adi, sh_ehalisi=None):
		self.ad = sh_adi
		self.ehali = sh_ehalisi
	def bomba_dushsun(self, nece_faiz_olsun):
		self.ehali = round(self.ehali * (1-nece_faiz_olsun))

sh1 = Sheher(sh_adi = 'baki', sh_ehalisi = 1e6)

print(sh1.ehali)
sh1.bomba_dushsun(0.8)
print(sh1.ehali)
```

```python
import math

class Cevre:
	def __init__(self, c_radiusu):
		self.radius = c_radiusu
	def sahe_tap(self):
		return math.pi * self.radius**2
	def perimetr_tap(self):
		return 2*math.pi * self.radius

c1 = Cevre(5)
c2 = Cevre(10)
c2.radius = 11
c1.perimetr_tap()
```
