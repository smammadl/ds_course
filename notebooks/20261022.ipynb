```python
class Circle:
	pi = 3.14

	def __init__(self, r):
		self.radius = r

	@property
	def diameter(self):
		return self.radius *2

	@property
	def area(self):
		return self.pi * (self.radius**2)

	@staticmethod
	def calculate_area(r):
		return Circle.pi * (r**2)

Circle.calculate_area(r=3)
```

```python
class Square:
	def __init__(self, a):
		self.side = a
	
	@property
	def area(self):
		return (self.side)**2

	@staticmethod
	def calculate_area(a):
		return a**2

# square1 = Square(a=5)
# square1.area
Square.calculate_area(a=127)
```

```python
class Cylinder:
	pi = 3.14

	def __init__(self, r, h):
		self.radius = r
		self.height = h

	@property
	def volume(self):
		return Cylinder.pi * (self.radius**2) * self.height

	@staticmethod
	def calculate_volume(r, h):
		return Cylinder.pi * r**2 * h

# slindr1 = Cylinder(r=2, h=5)
# slindr1.volume
Cylinder.calculate_volume(r=3, h=10)
```

```python
class Cylinder:
	n = 0
	pi = 3.14

	def __init__(self, r, h):
		self.radius = r
		self.height = h
		Cylinder.n = Cylinder.n + 1

	@property
	def volume(self):
		return Cylinder.pi * (self.radius**2) * self.height

	@staticmethod
	def calculate_volume(r, h):
		return Cylinder.pi * r**2 * h

	@classmethod
	def reset_n(cls):
		cls.n = 0

slindr1 = Cylinder(r=2, h=5)
slindr2 = Cylinder(r=3, h=2)
slindr3 = Cylinder(r=1, h=10)
print(Cylinder.n)
Cylinder.reset_n()
print(Cylinder.n)
```
