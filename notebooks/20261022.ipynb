```python
class Circle:
	pi = 3.14

	def __init__(self, r):
		self.radius = r

	@property
	def diameter(self):
		return self.radius *2

	@property
	def area(self):
		return self.pi * (self.radius**2)

	@staticmethod
	def calculate_area(r):
		return Circle.pi * (r**2)

Circle.calculate_area(r=3)
```

```python
class Cylinder:
	n = 0
	pi = 3.14

	def __init__(self, r, h):
		self.radius = r
		self.height = h
		Cylinder.n = Cylinder.n + 1

	@property
	def volume(self):
		return Cylinder.pi * (self.radius**2) * self.height

	@staticmethod
	def calculate_volume(r, h):
		return Cylinder.pi * r**2 * h

	@classmethod
	def reset_n(cls):
		cls.n = 0

slindr1 = Cylinder(r=2, h=5)
slindr2 = Cylinder(r=3, h=2)
slindr3 = Cylinder(r=1, h=10)
print(Cylinder.n)
Cylinder.reset_n()
print(Cylinder.n)
```

```python
class Rectangle:
	def __init__(self, a, b):
		self.length = a
		self.width = b

	@property
	def area(self):
		return self.length * self.width

	@staticmethod
	def calculate_area(a, b):
		return a*b

	@classmethod
	def calculate_perimeter(cls, a, b):
		return 2*a + 2*b
		

# r1 = Rectangle(a=10,b=5)
# r1.area
Rectangle.calculate_area(a=5, b=6)
Rectangle.calculate_perimeter(a=5, b=6)
```

```python
class Polygon:
	def __init__(self, number_of_sides):
		self.n = number_of_sides

class Triangle(Polygon):
	def __init__(self, angles_list):
		super().__init__(3)
		self.angles = angles_list

# t1 = Triangle([30,60,90])
# t1.angles

class Rectangle(Polygon):
	def __init__(self, a, b):
		super().__init__(4)
		self.length = a
		self.width = b

	@property
	def area(self):
		return self.length * self.width

class Square(Rectangle):
	def __init__(self, a):
		super().__init__(a, a)

r1 = Rectangle(a=2,b=3)
s1 = Square(a=3)
s1.n
```
