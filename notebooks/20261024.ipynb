```python
class Adam:
	def __init__(self, ad, doguldugu_il=None):
		self.ad = ad
		self.doguldugu_il = doguldugu_il
		self.yashayir = True

	def __del__(self):
		self.yashayir = False
		return 'adam oldu'

	def __repr__(self):
		return f"{self.__class__.__name__}({self.ad}, {self.doguldugu_il})"

	def __str__(self):
		return self.ad

	def __add__(self, other):
		return self.ad + other.ad

	def __sub__(self, other):
		if (self.doguldugu_il is not None) and (other.doguldugu_il is not None):
			return self.doguldugu_il - other.doguldugu_il
		elif (self.doguldugu_il is None) or (other.doguldugu_il is None):
			return None
		else:
			return int(self.doguldugu_il) - int(other.doguldugu_il)
	
	def __gt__(self, other):
		return (2026 - self.doguldugu_il) > (2026 - other.doguldugu_il)

a1 = Adam('ali', 1994)
a2 = Adam('veli', 1995)
a1
```

```python
from pprint import pprint

def qeydiyyatdan_kecir(ad, soyad, fin, *args, **kwargs):
	print(f"\'ad\' = {ad}")
	print(f"\'soyad\' = {soyad}")
	print(f"\'fin\' = {fin}")
	pprint(args, width=10)
	pprint(kwargs, width=10)

qeydiyyatdan_kecir(
	'ali', 
	'valiyev', 
	'12398kd', 
	'Imishli', 
	1995,
	yasadigi_yer='Baki', 
	ata_adi='hesen'
)
```

```python
def kvadrata_yukselt(herhansi_list):
	for nese in herhansi_list:
		kvadrat = nese**2
		return kvadrat

kvadrata_yukselt([-9,1,2,3,4,5,6])
```

```python
def kvadrata_yukselt(herhansi_list):
	for nese in herhansi_list:
		kvadrat = nese**2
		yield kvadrat

g = kvadrata_yukselt([-9,1,2,3,4,5,6])
```

```python
# 0,1,1,2,3,5,8,13,21

def nci_fibonaccini_tap(necenci):
	r1 = 0
	r2 = 1

	n = 2
	for i in range(necenci - 2):
		f = r1 + r2
		r1, r2 = r2, f
		n += 1

	return n, f

nci_fibonaccini_tap(19)
```

```python
# 0,1,1,2,3,5,8,13,21

def nci_fibonaccini_tap():
	r1 = 0
	r2 = 1

	n = 2
	while True:
		f = r1 + r2
		r1, r2 = r2, f
		n += 1

		yield n, f

fibonacci_generatoru = nci_fibonaccini_tap()
```
