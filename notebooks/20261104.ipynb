```python
for sayi in sayilar:
	beshe_bolunur = (sayi % 5 == 0)

	print(f'sayi: {sayi}')
	
	if beshe_bolunur:
		print(f'{sayi} beshe bolunur')
		break
```

```python
import random

while True:
	emr = input('emri daxil et: (yarat, cix)')
	taninan_emrler = ['yarat', 'cix', '0']

	if emr not in taninan_emrler:
		print(f'daxil etdiyiniz emr: {emr}, bele bir emr yoxdur')
		print(f'taninan emrler {taninan_emrler}')
		continue

	if emr == 'yarat':
		sayi = random.randint(0,100)
		print(sayi)
	elif emr == 'cix' or emr == '0':
		break
```
