```python
import numpy as np
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import fetch_openml

mnist = fetch_openml("mnist_784", as_frame=False)
mnist
```

```python
X = mnist['data']
X
```

```python
y = mnist['target']
np.unique(y, return_counts=True)
```

```python
fig,axs = plt.subplots(3,3)

for i, qutu in enumerate(axs.flatten()):
	X_reshaped = X[[i+100], :].reshape(28,28)
	qutu.imshow(X_reshaped, cmap='gray')
	qutu.set_title(y[i+100])
plt.show()
```

```python
pd.set_option("display.max_columns", None)

pd.DataFrame(X)
```

```python
X_scaled = X/255

from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.85, random_state=42)
X_train.shape, y_train.shape
```

```python
from sklearn.linear_model import LogisticRegression
log_reg = LogisticRegression()
log_reg.fit(X_train, y_train)
y_pred = log_reg.predict(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import VotingClassifier
vc = VotingClassifier([
		("log_reg", LogisticRegression()),
		("dec_tree", DecisionTreeClassifier()),
		("knn_cls", KNeighborsClassifier()),
	],
	voting = 'soft',
	weights=[1,2,1]
)

vc.fit(X_train, y_train)
```

```python
y_pred = vc.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_pred, y_test)
```

```python
from sklearn.linear_model import SGDClassifier
from sklearn.ensemble import BaggingClassifier
bc = BaggingClassifier(
	SGDClassifier(),
	n_estimators=3,
	max_samples=1000,
	oob_score=True
)
bc.fit(X_train, y_train)
```
