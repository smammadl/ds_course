```python
def get_mehsul_al(magazalar_mehsullar : dict):
	magazalar = list(magazalar_mehsullar.keys())
	mehsul_var = list(magazalar_mehsullar.values())

	for magaza_ind in range(len(magazalar)):

		magaza = magazalar[magaza_ind]
		magazada_mehsul_var = mehsul_var[magaza_ind]

		print(f'{magaza} magazaya girdim.')
		if magazada_mehsul_var:
			print('mehsul aldim')
			break
		else:
			print('mehsul yox idi, novbeti magazaya gedirem')

magazalar_mehsullar = {
	'A' : False,
	'A2' : False,
	'B' : True,
	'C' : False
}

get_mehsul_al(magazalar_mehsullar)
```

```python
magaza = {
	'corek' : {
		'qiymet' : 0.6,
		'stok' : 20
	},
	'su' : {
		'qiymet' : 0.8,
		'stok' : 30
	},
	'sud' : {
		'qiymet' : 1.5,
		'stok' : 10
	},
}

def inventar_goster(magaza : dict):
	for mehsul in magaza.keys():
		qiymet = magaza[mehsul]['qiymet']
		stok = magaza[mehsul]['stok']
		print(
				(
					f'{mehsul}: qiymet {qiymet} manat,'
					f' qalib {stok} eded'
				)
		)

inventar_goster(magaza)
```

```python
sebet = {
	'corek' : 2,
	'su' : 1
}

def odenish_ele(sebet):
	toplam_mebleg = 0
	for mehsul, eded in sebet.items():
		qiymet = magaza[mehsul]['qiymet']

		toplam = eded * qiymet
		print(f'{eded} eded {mehsul}: {toplam} manat')
		
		toplam_mebleg += toplam

		magaza[mehsul]['stok'] -= eded

	print(f"toplam mebleg: {toplam_mebleg} manat")
	return toplam_mebleg

mebleg = odenish_ele(sebet)
```

```python
magaza = {
	'corek' : {
		'qiymet' : 0.6,
		'stok' : 20
	},
	'su' : {
		'qiymet' : 0.8,
		'stok' : 30
	},
	'sud' : {
		'qiymet' : 1.5,
		'stok' : 10
	},
}

def mehsul_haqqinda_melumat_ver(mehsul):
	melumat : dict = magaza.get(mehsul)

	if melumat:
		print(melumat)
	else:
		melumat = {
			'qiymet' : None,
			'stok' : None
		}
		print(melumat)
	return melumat

melumat = mehsul_haqqinda_melumat_ver('corek')
```

```python
emr = 'melumat corek'

l = emr.split(' ')
mehsul = l[1]
mehsul
```

```python
while True:
	emr = input('emr daxil edin: (inventar, melumat <mehsul>, cix)')

	if emr == 'inventar':
		inventar_goster(magaza)
	elif 'melumat' in emr:
		l = emr.split(' ')
		mehsul = l[1]
		mehsul_haqqinda_melumat_ver(mehsul)
	elif emr == 'cix':
		print('proqram bitdi')
		break
	else:
		print('bele bir emr tapilmadi')
```
