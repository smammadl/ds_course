```python
HAZIRKI_IL = 2026
axtarilan_ad = None
axtarilan_soyad = None

tapilan_musteriler = []

for musteri_no in list(musteriler.keys()):
	musteri_ad = musteriler[musteri_no]['ad']
	musteri_soyad = musteriler[musteri_no]['soyad']
	musteri_tel = musteriler[musteri_no]['telefon']
	musteri_yash = HAZIRKI_IL - musteriler[musteri_no]['dogum_tarixi']

	if (
		(musteri_ad.lower() == axtarilan_ad.lower() if axtarilan_ad else True)
		and 
		(musteri_soyad.lower() == axtarilan_soyad.lower() if axtarilan_soyad else True)
	):
		print(f'{musteri_no} nomreli musteri')
		print(f'ad:{musteri_ad}, soyad:{musteri_soyad}, tel:{musteri_tel}, yash:{musteri_yash}')
		print()
		tapilan_musteriler.append({
			musteri_no : musteriler[musteri_no]
		})
```
