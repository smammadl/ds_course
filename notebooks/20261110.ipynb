```python
# df.groupby('Make').agg({
# 	'Price' : 'max',
# 	'Odometer (KM)' : 'min'
# })
df_g = df.groupby('Make').agg(
	Price_max = ('Price', 'max'),
	Price_min = ('Price', 'min'),
)

df_g['Diff'] = df_g['Price_max'] - df_g['Price_min']
df_g
```

```python
def bahali_ucuz(x):
	if x.mean() > 20_000:
		return 'bahali'
	else:
		return 'ucuz'

df.groupby("Make")['Price'].apply(bahali_ucuz)
```

```python
# df.groupby("Make")['Doors'].apply(pd.Series.mode)

def mode_tap(x):
	return x.mode()[0]

df.groupby("Make")['Doors'].apply(mode_tap)
```

```python
df.groupby("Make")['Doors'].apply(lambda x: x.mode()[0])
```

```python
# def bahali_ucuz(x):
# 	if x.mean() > 20_000:
# 		return 'bahali'
# 	else:
# 		return 'ucuz'

df.groupby("Make")['Price'].apply(lambda x: 'bahali' if x.mean() > 20_000 else 'ucuz')
```

```python
# df['Make'].value_counts(dropna=False)
df.dropna(subset='Make', inplace=True)
```

```python
price_mean = df.groupby('Make')['Price'].transform('mean')

df['Price_mean'] = price_mean
df
```

```python
# def al_gozle(x):
# 	return 'al' if x['Price'] < x['Price_mean'] else 'gozle'

# qerar = df[['Price','Price_mean']].apply(al_gozle, axis=1)

qerar = df[['Price','Price_mean']].apply(
	lambda x : 'al' if x['Price'] < x['Price_mean'] else 'gozle', 
	axis=1
	)
df['Decision'] = qerar
df
```
