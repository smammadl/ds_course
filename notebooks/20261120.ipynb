```python
from pathlib import Path
BASE_DIR = Path.cwd()
BANKLAR_DIR = BASE_DIR / 'banklar'
BANKLAR_DIR.mkdir(exist_ok=True)

class Bank:
	def __init__(self, b_adi):
		self.ad = b_adi
		self.hesablar = []
		self.dir = BANKLAR_DIR / b_adi
		self.dir.mkdir(exist_ok=True)
		self.hesablar_csv = self.dir / 'hesablar.csv'
		with open(self.hesablar_csv, 'w') as f:
			f.write('fin,nomre,valyuta')

	def __repr__(self):
		return f"<{self.__class__.__name__}: {self.ad}, {len(self.hesablar)}>"

	def bankin_hesablarini_oxu(self):
		with open(self.hesablar_csv, 'r') as f:
			lines = f.readlines()
			mene_lazim_olan_lines = lines[1:]
		return mene_lazim_olan_lines

	def hesab_nomrelerinin_en_boyuyunu_tap(self, hesablar):
		if not hesablar:
			return 1
		else: 
			max([int(hesab[1]) for hesab in hesablar])

	def hesab_yarat(self, h_sahibinin_finkodu, h_valyutasi):
		hesablar = self.bankin_hesablarini_oxu()
		max_hesab = self.hesab_nomrelerinin_en_boyuyunu_tap(hesablar)
		yeni_hesab_nomresi = max_hesab + 1
		h = Hesab(h_sahibinin_finkodu, yeni_hesab_nomresi, h_valyutasi)
		self.hesablar.append(h)
		return h

class Hesab:
	def __init__(self, h_sahibinin_finkodu, h_nomresi, h_valyutasi):
		self.sahibin_finkodu = h_sahibinin_finkodu
		self.valyuta = h_valyutasi
		self.nomre = h_nomresi

	def finkodu_ulduzla(self):
		return self.sahibin_finkodu[0] + '*'*5 + self.sahibin_finkodu[-1]

	def __repr__(self):
		return f"Hesab: {self.finkodu_ulduzla()}, {self.valyuta}"


b1 = Bank('abb')
b1 = Bank('kapital')
x = b1.hesab_yarat('435LA09', 'AZN')
x = b1.hesab_yarat('afasdfa', 'EUR')
```
