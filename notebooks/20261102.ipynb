```python
contacts = {
	'ali' : {
		'ad soyad' : 'Ali Hesenov', 
		'telefon' : '+994551234567',
		'dogum tarixi' : '1995',
	},
	'hesen' : {
		'ad soyad' : 'Hesen Huseynov', 
		'telefon' : '+994551234568',
		'dogum tarixi' : '1996',
	},
	'lale' : {
		'ad soyad' : 'Lalə Əliyeva', 
		'telefon' : '+994501234569',
		'dogum tarixi' : '1997',
	},
}

adam = input('adamin adini daxil et: ')
adam_listedir = adam in contacts.keys()

if adam_listedir:
	adamin_ad_soyadi = contacts[adam]['ad soyad']
	adamin_telefon_nomresi = contacts[adam]['telefon']
	adamin_yashi = 2026 - int(contacts[adam]['dogum tarixi'])

	print(f"{adamin_ad_soyadi}: mobil {'0' + adamin_telefon_nomresi[4:]}, yash: {adamin_yashi}")
else:
	print(f'{adam} adinda adam yoxdur')
```

```python
magaza = {
	"sud" : {
		"qiymet" : 2.3,
		"stok" : 10,
	},
	"corek" : {
		"qiymet" : 0.6,
		"stok" : 30,
	},
	"kolbasa" : {
		"qiymet" : 4,
		"stok" : 5,
	},
}

ad = 'corek'
say = 10

magazada_mehsul_var = ad in list(magaza.keys())

stokdaki_say = int(magaza[ad]['stok'])
stokda_yeterli_mehsul_var = (say <= stokdaki_say)

if magazada_mehsul_var:
	if stokda_yeterli_mehsul_var:
		qiymet = magaza[ad]['qiymet']
		mebleg = float(say) * float(qiymet)

		print(f'{ad}: {say} eded x {qiymet} manat = {mebleg} manat')

		magaza[ad]['stok'] = magaza[ad]['stok'] - say
		print(f"stokda {magaza[ad]['stok']} eded {ad} qaldi")

	else:
		print(f'stokda {say} {ad} mehsulu yoxdur')
else:
	print('bele mehsul yoxdur')
```
