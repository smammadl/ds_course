```python
(
	dfs['employees']
	.merge(dfs['dept_emp'], on='emp_no', how='left')
	.merge(dfs['departments'], on='dept_no', how='left')
	.loc[:, ['emp_no','first_name','last_name','dept_no','dept_name']]
	.groupby(by='emp_no')['dept_no'].count()
	.sort_values(ascending=False)
)
```

```python
(
	dfs['departments']
	.merge(dfs['dept_emp'], on='dept_no', how='left')
	.merge(dfs['salaries'], on='emp_no', how='left')
	# .drop(['from_date_x', 'to_date_x', 'from_date_y', 'to_date_y'], axis=1)
	.loc[:, ['dept_no', 'dept_name', 'emp_no', 'salary']]
	.groupby(by=['dept_no', 'dept_name', 'emp_no'])['salary'].max().to_frame().reset_index()
	.groupby(by=['dept_no'])['salary'].mean()
	.sort_values(ascending=False)
)
```
